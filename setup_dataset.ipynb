```python
#Selecting only 5 geners
import os
import shutil

# List of genres you want to include
selected_genres = ['blues', 'classical', 'rock', 'hiphop', 'jazz']

def filter_audio_files(src_dir, dest_dir, selected_genres):
    for genre in selected_genres:
        genre_dir = os.path.join(src_dir, genre)
        if os.path.isdir(genre_dir):
            # Create subfolder for the selected genre in the destination folder
            os.makedirs(os.path.join(dest_dir, genre), exist_ok=True)
            
            # Copy all audio files for the selected genre to the destination folder
            for file in os.listdir(genre_dir):
                if file.endswith('.wav'):
                    shutil.copy(os.path.join(genre_dir, file), os.path.join(dest_dir, genre, file))

# Example usage
src_audio_dir = 'E:/College Mtech/Sem 2/AMLDL Lab/8 Mini Project/Mini Project/Data/genres_original'  # Original folder with all genres
dest_audio_dir = 'E:/College Mtech/Sem 2/AMLDL Lab/8 Mini Project/Mini Project/dataset/filtered_audio'  # New folder for filtered genres
filter_audio_files(src_audio_dir, dest_audio_dir, selected_genres)
```

```python
#Splitting audio data in train and val
import os
import shutil
from sklearn.model_selection import train_test_split

def split_audio_data(src_dir, dest_dir, val_size=0.2):
    genres = os.listdir(src_dir)  # List of genres (subfolders in the src_dir)
    
    for genre in genres:
        genre_dir = os.path.join(src_dir, genre)
        if os.path.isdir(genre_dir):
            # Get all audio files in the genre folder
            audio_files = [f for f in os.listdir(genre_dir) if f.endswith('.wav')]
            
            # Split into train and validation (80% train, 20% validation)
            train_files, val_files = train_test_split(audio_files, test_size=val_size, random_state=42)
            
            # Create genre subfolders in train and val
            os.makedirs(os.path.join(dest_dir, 'train', genre), exist_ok=True)
            os.makedirs(os.path.join(dest_dir, 'val', genre), exist_ok=True)
            
            # Copy train audio files
            for file in train_files:
                shutil.copy(os.path.join(genre_dir, file), os.path.join(dest_dir, 'train', genre, file))
                
            # Copy validation audio files
            for file in val_files:
                shutil.copy(os.path.join(genre_dir, file), os.path.join(dest_dir, 'val', genre, file))

# Example usage
src_audio_dir = 'E:/College Mtech/Sem 2/AMLDL Lab/8 Mini Project/Mini Project/dataset/filtered_audio'  # Folder containing genre subfolders with audio files (.wav)
dest_audio_dir = 'E:/College Mtech/Sem 2/AMLDL Lab/8 Mini Project/Mini Project/dataset/split_audio'  # Destination folder to save train and val split
split_audio_data(src_audio_dir, dest_audio_dir)
```

```python
#spliting spectograms into train and validation subfolders 

import os
import shutil
from sklearn.model_selection import train_test_split

def split_images(src_dir, dest_dir, val_size=0.2):
    genres = os.listdir(src_dir)
    
    for genre in genres:
        genre_dir = os.path.join(src_dir, genre)
        if os.path.isdir(genre_dir):
            # Get all image files in the genre folder
            image_files = [f for f in os.listdir(genre_dir) if f.endswith(('.png', '.jpg', '.jpeg'))]
            
            # Split into train and validation
            train_files, val_files = train_test_split(image_files, test_size=val_size, random_state=42)
            
            # Create genre subfolders in train and val
            os.makedirs(os.path.join(dest_dir, 'train', genre), exist_ok=True)
            os.makedirs(os.path.join(dest_dir, 'val', genre), exist_ok=True)
            
            # Copy train images
            for file in train_files:
                shutil.copy(os.path.join(genre_dir, file), os.path.join(dest_dir, 'train', genre, file))
                
            # Copy validation images
            for file in val_files:
                shutil.copy(os.path.join(genre_dir, file), os.path.join(dest_dir, 'val', genre, file))

# Example usage
src_images_dir = 'E:/College Mtech/Sem 2/AMLDL Lab/8 Mini Project/Mini Project/dataset/filtered_spectrograms'  # Folder containing genre subfolders with images
dest_images_dir = 'E:/College Mtech/Sem 2/AMLDL Lab/8 Mini Project/Mini Project/dataset/split_images'  # Destination folder to save train and val split
split_images(src_images_dir, dest_images_dir)
```
